# Data Cleaning + Handling TotalCharges, Missing Values, and Column Separation


```python
import pandas as pd
import numpy as np
```

```python
# Create a copy of the original dataset
df = pd.read_csv(
    r"C:\Users\infor\OneDrive\Desktop\customer_churn_prediction\data\raw\Telco-Customer-Churn.csv"
)
df_clean = df.copy()
# Show first 5 rows
df_clean.head()
```

```python
print("Shape before cleaning:", df_clean.shape)
```

## Drop Unnecessary Column `customerID`


```python
# Drop customerID column because it is only an identifier
df_clean.drop("customerID", axis=1, inplace=True)

# Check shape after dropping
df_clean.shape
```

## Inspect `TotalCharges`


```python
# Check first few values of TotalCharges
df_clean["TotalCharges"].head(10)
```

```python
# Check values where TotalCharges contains blank spaces
df_clean[df_clean["TotalCharges"] == " "]
```

## Convert `TotalCharges` to Numeric


```python
# Convert TotalCharges to numeric
# errors="coerce" converts invalid values like blank spaces into NaN
df_clean["TotalCharges"] = pd.to_numeric(df_clean["TotalCharges"], errors="coerce")
```

```python
# Check missing values after conversion TotalCharges
df_clean.isnull().sum()
```

```python
# Show rows where TotalCharges is missing
df_clean[df_clean["TotalCharges"].isnull()]
```

```python
# Drop rows where TotalCharges is missing
df_clean.dropna(subset=["TotalCharges"], inplace=True)

# Check missing values again
df_clean.isnull().sum()
```

```python
# Reset index after dropping rows
df_clean.reset_index(drop=True, inplace=True)

# Check first 5 rows
df_clean.head()
```

## Check Duplicate Rows


```python
# Check duplicate rows
duplicate_count = df_clean.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
```

```python
# Remove duplicate rows if any
df_clean.drop_duplicates(inplace=True)

# Check shape after removing duplicates
print("Shape after removing duplicates:", df_clean.shape)
```

## Convert Target Column `Churn`

Yes -> 1

No -> 0


```python
# Convert target column into numeric values
df_clean["Churn"] = df_clean["Churn"].map({"Yes": 1, "No": 0})

# Check target values
df_clean["Churn"].value_counts()
```

0 = customer stayed

1 = customer churned


```python
# Check Data Types After Cleaning
df_clean.dtypes
```

```python
# Final missing value check
df_clean.isnull().sum()
```

## Separate Features and Target

X = input features

y = target/output


```python
# Separate input features and target column
X = df_clean.drop("Churn", axis=1)
y = df_clean["Churn"]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

## Identify Numerical and Categorical Columns


```python
# Identify numerical columns
numerical_columns = X.select_dtypes(include=["int64", "float64"]).columns.tolist()

# Identify categorical columns
categorical_columns = X.select_dtypes(include=["str"]).columns.tolist()

print("Numerical columns:")
print(numerical_columns)

print("\nCategorical columns:")
print(categorical_columns)
```

```python
for col in categorical_columns:
    print(f"\nColumn: {col}")
    print(df_clean[col].unique())
    print("Number of unique values:", df_clean[col].nunique())
```

```python
# Save cleaned dataset
df_clean.to_csv("../data/processed/Telco-Customer-Churn-Cleaned.csv", index=False)
```
